```python
class TourPlanner:
    def __init__(self, company_name):
        # Initialize the company name
        self.company_name = company_name
        
        # Define a list of dictionaries for tour packages
        # These are stored as an instance variable
        self.tour_packages = [
            {'name': 'Mountain Escape', 'price': 200},
            {'name': 'Beach Paradise', 'price': 150},
            {'name': 'City Lights', 'price': 100},
            {'name': 'Desert Safari', 'price': 250}
        ]

    def book(self, person_name, number_of_people, package_name):
        """
        Handles the booking logic and returns the bill details.
        """
        # Search for the package in our list
        selected_package = None
        for package in self.tour_packages:
            if package['name'].lower() == package_name.lower():
                selected_package = package
                break
        
        if selected_package:
            # Calculate the total bill
            total_price = selected_package['price'] * number_of_people
            return self.generate_bill(person_name, number_of_people, selected_package['name'], total_price)
        else:
            return f"Sorry, the package '{package_name}' was not found."

    def generate_bill(self, person_name, number_of_people, package_name, total_price):
        """
        Formats and provides the bill details.
        """
        bill_info = (
            f"\n--- {self.company_name} Official Bill ---\n"
            f"Customer Name: {person_name}\n"
            f"Package: {package_name}\n"
            f"Number of People: {number_of_people}\n"
            f"Total Amount: ${total_price}\n"
            f"-----------------------------------"
        )
        return bill_info

# --- Example Usage ---
# 1. Create the object
my_planner = TourPlanner("Globetrotter Adventures")

# 2. Book a trip
receipt = my_planner.book("Alice Smith", 4, "Mountain Escape")

# 3. Print the result
print(receipt)
```
